- **Tranchage d'un tableau :**<br>
Obtenir et créer de petits sous-tableaux à partir d'un grand tableau

>**Sous tableau à une dimension**



```python
# x[start:end:step]
print(x1)
x1[:3] # les trois premier élément
```

```python
x1[3:] # à partir de l'index 3
```

```python
x1[3:5]
```

```python
x1[::2] # pas de 2
```



>**Sous-tableau à plusieurs dimension**


```python
#La tranche séparer par des virgule
print(x3,"\n")
print("séparation \n")
x3[:2,:2,:3] # récuperation de la première et deuxième dimension, recuperation des deux premières lignes et trois colonnes
```

```python
# note des étudiants
arr_note = np.array([12,11,9,18,6,17,8])
```

```python
print("la moyenne :", arr_note.mean())
print("minimum:", arr_note.min())
print("maximum:", arr_note.max())
print("nombre étudiant ayant la moyenne", len(arr_note[arr_note>=arr_note.mean()]))# methode avec len
print("nomre étudiant ayant inferieure à la moyenne générale", np.sum(arr_note>arr_note.mean())) # methode avec sum de numpy
```

```python
# Change les notes inférieure à 10 en leur faisant plus 2
arr_note[arr_note<10] = arr_note[arr_note<10] + 2
arr_note
```

```python
print("nombre étudiant ayant la moyenne", len(arr_note[arr_note>=10]))
```
