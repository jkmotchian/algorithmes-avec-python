Mode debogage avec %xmode ayant trois option : Plain, context et verbose

```python
def fonc1(a,b):
  return a/b

def fonc2(x):
  a = x
  b = x-1
  return fonc1(a,b)

fonc2(1)
```

```python
%xmode Plain
# mode compact
```

```python
fonc2(1)
```

```python
%xmode VERBOSE # mode plus bavard que les deux autres
fonc2(1)
```

```python
fonc2(1)
```

```python
%debug
```

## **Profilage et chronométrage d'exécution**

il est important de finir le code entier avant de commencer a optimiser le code et non a chaque ligne de code on cherche à optimiser.<br>
Pour obtenir le temps d'exécution d'une fonction on pour utiliser les commandes magiques

```python
# commande %timeit et %%timeit

%%timeit
Total = 0
for i in range (1000):
  for j in range (1000):
    Total += i*(-1)**j
```

```python
# commande %timeit et %%timeit
%timeit
Total = 0
for i in range (1000):
  for j in range (1000):
    Total += i*(-1)**j
```

```python
import random
L = [random.random() for i in range(100000)]
print ("Tri d'une liste non triée")
%time L.sort()
```

```python
L= [random.random () for i in range (10000)]
print ("Tri d'une liste non triée")
%timeit L.sort()
```

```python
def sommer_listes(N):
  total = 0
  for i in range (5):
    L= [j^(j>>i) for j in range (N)]
    total += sum(L)
    return total
```

Pour profiler de façon globale une fonction on peut utiliser %prun

```python
%prun sommer_listes(10000)
```
